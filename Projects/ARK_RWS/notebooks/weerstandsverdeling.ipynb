# Eenvoudige weerstand checker

Hoeveel kan de stijghoogte naast het ARK omhoog gedurende de zomer als gevolg van de hogere temperatuur van het kanaalwater.

Het idee is om een stroomlijn te beschouwen met debieet $q$ tussen het kanaal en een sloot naast het kanaal. Het traject wordt opgedeeld in een aantal deeltrajecten die elk een eigen weerstand hebben. Het totale stijghoogteverschil tussen kanaal en sloot naast het kanaal ligt vast. De weerstanden ook, maar die zijn afhankelijk van de viscositeit. Geschikte cijfers zijn af te leiden uit de numerieke doorrekening van de doorsnede.

We hebben $\phi_0$ en $\phi_p$ het kanaalpeil en het polderpeil. En de volgende weerstanden $R1$ kanaalbodem, $R2$ tussen kanaalbodem en onderzijde damwanden, $R3# tussen onderzijde damwanden en onderzijde deklaag en $R4$ tussen onderzijde deklaag en sloot.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
phiARK, phiP = -0.4, -2.4
Phi =np.array([phiARK, 0, 0, 0, phiP])
x = np.arange(len(Phi))
R = np.ones(len(Phi)-1)

fig, ax = plt.subplots()

for temp in [5, 10, 15, 20, 25]:
    mu = 30 / (temp + 20)
    R[0] = mu

    q = (phiARK - phiP) / np.sum(R)

    for i, r in enumerate(R):
        Phi[i + 1] = Phi[i] * r * q
    
    ax.plot(x, Phi, 'o-', label=f'T={temp}')

ax.grid()
ax.legend()

```
