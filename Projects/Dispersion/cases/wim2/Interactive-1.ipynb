```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma, lognorm

# Some positive-skewed data (e.g., hydraulic conductivity)
data = np.random.lognormal(mean=0.5, sigma=0.9, size=1000)

# Fit both distributions
shape_g, loc_g, scale_g = gamma.fit(data)
shape_l, loc_l, scale_l = lognorm.fit(data)

x = np.linspace(min(data), max(data), 1000)
pdf_gamma = gamma.pdf(x, shape_g, loc=loc_g, scale=scale_g)
pdf_lognorm = lognorm.pdf(x, shape_l, loc=loc_l, scale=scale_l)

plt.hist(data, bins=50, density=True, alpha=0.3, label="Data")
plt.plot(x, pdf_gamma, 'r-', label="Pearson III (Gamma)")
plt.plot(x, pdf_lognorm, 'g--', label="Lognormal")
plt.legend()
plt.title("Comparison of Lognormal vs Pearson III (Gamma)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma, lognorm

# Example data: skewed, positive-only (e.g. travel times or K values)
data = np.random.lognormal(mean=0.5, sigma=0.9, size=1000)

# Fit the distributions
shape_g, loc_g, scale_g = gamma.fit(data)
shape_l, loc_l, scale_l = lognorm.fit(data)

# Range of values for plotting
x = np.linspace(min(data), np.percentile(data, 99.9), 1000)

# Compute CDFs
cdf_gamma = gamma.cdf(x, shape_g, loc=loc_g, scale=scale_g)
cdf_lognorm = lognorm.cdf(x, shape_l, loc=loc_l, scale=scale_l)

# Empirical CDF from data
data_sorted = np.sort(data)
empirical_cdf = np.arange(1, len(data_sorted)+1) / len(data_sorted)

# Plot
plt.plot(data_sorted, empirical_cdf, 'k.', label='Empirical CDF')
plt.plot(x, cdf_gamma, 'r-', label='Pearson III (Gamma) CDF')
plt.plot(x, cdf_lognorm, 'g--', label='Lognormal CDF')
plt.xlabel("Value")
plt.ylabel("Cumulative Probability")
```

```python
# pass_times dtype([('id', '<i4'), ('xObs', '<f8', (10,)), ('time', '<f8', (10,))])
# pass_times has the times of each particle (id) for a set of x-locations
pass_times = np.load(os.path.join(dirs.data, 'pass_times_' + case_name + '.npy'))

xObs = pass_times['xObs'][0]
nxObs = len(xObs)

# Get just the x and times so we can sort the times (eliminate the ids)
dtype0t = np.dtype([('xObs', '<i4', (nxObs,)), ('time', '<f8', (nxObs,))])
cumul_t = np.zeros(len(pass_times), dtype=dtype0t)  # dict()
cumul_t['xObs'] = pass_times['xObs']

# Sort the times
for j in range(nxObs):
    idx = np.argsort(pass_times['time'][:, j])    
    t = pass_times['time'][:, j][idx]    
    cumul_t['time'][:, j] = t
    
# Shorten the xlim for better picture
last_times = cumul_t['time'][:, -1]
t_last = last_times[int(0.95 * len(last_times))]

tlim = (0, round_up_nice(t_last, m=2))
```
