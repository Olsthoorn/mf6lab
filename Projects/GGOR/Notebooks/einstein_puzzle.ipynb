# Einstein puzzles

Einstein puzzels zijn logische puzzels waarin je op basis van in zinnen verpakte gegevens unike combinaties moet bepalen van een aantal categoriëen.

Vragen:

1. Anna heeft niet de hond.
2. Wie de kat heeft, graag een groene trui.
3. Bram draagt geen blauwe trui.
4. Clara heeft de vis.

```python
wie = ['A', 'B', 'C']
dier = ['K', 'H', 'V']
kleur = ['B', 'G', 'R']
```

```python
for w in wie:
    print(w, end="")
    for d in dier:
        print(d, end="")
        for k in kleur:
            print(k)
            if not ((w == 'A') and (not (d == 'H'))):
                break
            if not ((d == 'K') and (k == 'G')):
                break
            if not((w == 'B') and (not (k == 'B'))):
                break
            if not ((w == 'C') and (d == 'V')):
                break
            print(w, d, k)
```
