Partial penetration of galleries (Huisman 1972) p55

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
b = 250
kD = 500
p = 0.001
x = np.linspace(0, b, 101)
h = p/(2 * kD) * (b**2 - x**2)
havg = p/(3 * kD) * b**2

title2 = f"p={p} m/d, kD={kD} m2/d, b={b} m"

fig, ax = plt.subplots()
ax.set_title("Stationaire oplossing Dupuit en gemiddelde grondwaterstand"
             "\n" + title2)
ax.set(xlabel='x [m]', ylabel='h [m]')
ax.plot(x, h)
ax.axhline(havg, 0, b)
ax.plot(b / np.sqrt(3), havg, 'ro', label=r"$x(h_x=\overline{h})=b/\sqrt{3} \approx 0.58b$")
ax.grid(True)
print(1/np.sqrt(3))
ax.legend(loc='lower left')

parts = list(Path(os.getcwd()).parts)
pth = os.path.join(*parts[:parts.index('GGOR') + 1], 'doc', 'images')
fig.savefig(os.path.join(pth,"dupuit_stat_avg.png"))
print(os.path.isdir(pth))
Path
```
